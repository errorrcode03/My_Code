```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer as cv
vector=cv()

corpus = [
    "The product is amazing. I really enjoyed using it!",
    "The delivery was delayed. I was waiting for three days.",
    "The headphones are comfortable and working perfectly.",
    "I purchased this laptop yesterday. It performs extremely well.",
    "The battery drains quickly. I am thinking about returning it."
] 
op=vector.fit_transform(corpus)
op.toarray
```

```python
frame=pd.DataFrame(op)
frame
```
