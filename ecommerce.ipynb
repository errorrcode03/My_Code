```python
import pandas as pd 
link="S:\\ML\\Kaggle_dataset\\e_commerce\\ecommerce_orders_dataset.csv"
df=pd.read_csv(link)
df.head()
```

```python
x=df[["Order_Amount","Tax_Amount","Shipping_Cost","Unit_Price","Quantity"]]
y=df["Profit_Margin_Percent"]
```

```python
from sklearn.neighbors import KNeighborsRegressor as knr
model=knr(n_neighbors=1)
from sklearn.model_selection import train_test_split as ttp
x_train,x_test,y_train,y_test=ttp(x,y,test_size=0.12,random_state=12)
model.fit(x_train,y_train)
pred=model.predict(x_test)

import seaborn as sns
sns.displot(pred-y_test)
```
