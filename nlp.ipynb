```python
import pandas as pd 
df=pd.DataFrame({"corpus":["People watch porn","People visits xvideos ","xvideos has porn","Japani do handle daily"]})
df
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer as tfv
vector=tfv()
text=vector.fit_transform(df["corpus"])
print(text.toarray())
```

```python
from sklearn.feature_extraction.text import CountVectorizer as cv
count=cv()

frq=count.fit_transform(df["corpus"])

print(count.vocabulary_)
print(frq.toarray())
```
