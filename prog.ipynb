```python
from sklearn.datasets import fetch_california_housing
import pandas as pd
masirchop=fetch_california_housing()
dataset=pd.DataFrame(masirchop.data)
dataset.columns=masirchop.feature_names
x=dataset
y=masirchop.target
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.15,random_state=10)
from sklearn.preprocessing import StandardScaler
errorcode=StandardScaler()
x_train=errorcode.fit_transform(x_train)
x_test=errorcode.transform(x_test)
from sklearn.neighbors import KNeighborsRegressor as Kn
chutt= Kn(n_neighbors=5)
chutt.fit(x_train,y_train)
paglu=chutt.predict(x_test)
print(paglu[0].round(2)*2)
import seaborn as sns
sns.displot(paglu-y_test)
import numpy as np 
mape = np.mean(np.abs((y_test-paglu)/y_test))*100
print("Accuracy :",100-mape.round(2),"\n Mape :",mape.round(2))
```
