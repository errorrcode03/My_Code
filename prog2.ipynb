```python
from sklearn.datasets import fetch_california_housing 
import numpy as np
import pandas as pd 
obj1=fetch_california_housing()
dataset=pd.DataFrame(obj1.data)
dataset.columns=obj1.feature_names
x=dataset
y=obj1.target
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.30,random_state=20)
from sklearn.preprocessing import StandardScaler
std=StandardScaler()
x_train=std.fit_transform(x_train)
x_test=std.transform(x_test)
from sklearn.linear_model import LinearRegression 
dipam=LinearRegression()
dipam.fit(x_train,y_train)
pred=dipam.predict(x_test)
from sklearn.model_selection import cross_val_score
errorr= cross_val_score(dipam,x_train,y_train,scoring='neg_mean_squared_error',cv=20)
print(f"training accuracy : {np.mean(errorr):.5f}")
print(pred)
import seaborn as sns
sns.displot(pred-y_test)
```
