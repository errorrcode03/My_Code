```python
from sklearn.datasets import _california_housing, fetch_california_housing
import pandas as pd 
fetching = fetch_california_housing()
framing= pd.DataFrame(fetching.data)
cols= fetching.feature_names
print(cols)
print(framing.round(2))
x=framing
y=fetching.target
print(y.round(1))
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test= train_test_split(x,y,test_size=0.40,random_state=20,shuffle=True)
from sklearn.preprocessing import StandardScaler
scale= StandardScaler()
x_train=scale.fit_transform(x_train)
x_test=scale.transform(x_test)
from sklearn.linear_model import LinearRegression
mode= LinearRegression()
mode.fit(x_train,y_train)
predicts=mode.predict(x_test)
print(predicts.round(2),"\n Total Size :",predicts.size)
import seaborn as sns
sns.displot(predicts-y_test)
import matplotlib.pyplot as plt 
plt.scatter(predicts,y_test)
plt.xlabel("Predicted Values")
import numpy as np 
mape = np.mean((np.abs(y_test-predicts)/y_test))*100
print("MAPE :",mape.round(2))
```
