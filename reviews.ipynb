```python
from nltk.tokenize import word_tokenize 

corpus = [
    "The product is amazing. I really enjoyed using it!",
    "The delivery was delayed. I was waiting for three days.",
    "The headphones are comfortable and working perfectly.",
    "I purchased this laptop yesterday. It performs extremely well.",
    "The battery drains quickly. I am thinking about returning it."
] 
```

```python
reviews=[]
for i in corpus:
    reviews.append(word_tokenize(i))

reviews
```

```python
import nltk
from nltk.corpus import stopwords 
nltk.download('stopwords')
search=set(stopwords.words('english'))
from nltk.stem import RegexpStemmer 
regular=RegexpStemmer('ing$',min=3)
from nltk.stem import PorterStemmer 
harry=PorterStemmer()
from nltk.stem import SnowballStemmer 
ball=SnowballStemmer('english')
```

```python

tokens=[]
count=0
for string in reviews:
           temp=[]
           if count==0:
                   for i in string:
                           if i not in search:
                                temp.append(i)
           elif count ==1:
                 for i in string:
                        temp.append(harry.stem(i))  
           elif count==2:
                   for i in string:
                           temp.append(ball.stem(i))       
           elif count>=3:
                   for i in string:
                           temp.append(regular.stem(i))

           count+=1
           tokens.append(temp)

tokens
```
