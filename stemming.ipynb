```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.tokenize import sent_tokenize
corpus="I cannot do maths. It is too hard. I am going to school . I like to study computer science . I like DSA a lot. "
docs=sent_tokenize(corpus)
words=word_tokenize(corpus)
print("Sentences :",docs)
print("Words :",words)
```

```python
from nltk.corpus import stopwords
nltk.download('stopwords')
sw=set(stopwords.words('english'))
filter=[]
for i in words:
    if i not in sw:
        filter.append(i)

print(filter)
```

```python
from nltk.stem import PorterStemmer as ps 
stemming= ps()
dick={}
for i in filter:
    dick[i]=stemming.stem(i)

print(dick)
```

```python
from nltk.stem import RegexpStemmer as regular_expression
exp=regular_expression('ing$',min=3)
dick2={}
for i in filter:
    dick2[i]=exp.stem(i)

print(dick2)
```

```python
from nltk.stem import SnowballStemmer as balls
best_stem=balls('english')
text=" My city has a historical rich background. "
dummy=word_tokenize(text)
hashmap={}
for i in dummy:
    hashmap[i]=best_stem.stem(i)

print(hashmap)


```
