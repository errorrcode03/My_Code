```python
import pandas as pd 
link = "S:\\ML\\Kaggle_dataset\\Customer Data.csv"
dataset=pd.read_csv(link)
dataset.head()
```

```python
from sklearn.cluster import KMeans
model=KMeans(n_clusters=5,random_state=23)
x=dataset.drop("CUST_ID",axis=1)
from sklearn.impute import SimpleImputer 
clean=SimpleImputer()
x=clean.fit_transform(x)
model.fit(x)
output=model.labels_
```

```python
from collections import Counter 
hash=Counter(output)
print(hash)
```
